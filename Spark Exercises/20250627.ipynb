```python
!pip install pyspark
from pyspark import SparkConf, SparkContext
conf = SparkConf().setAppName("Prova_esame")
sc = SparkContext(conf=conf)
```

```python
outputPath1="./output1/"
productsRDD=sc.textFile("./data/Products.txt")
purchasesRDD=sc.textFile("./data/Purchases.txt")
usersRDD=sc.textFile("./data/Users.txt")
```

#Task 1

```python
cleanedPurchasesRDD=purchasesRDD.map(lambda x: (x.split(",")[1],x.split(",")[3])) \
    .filter(lambda x: int(x[1].split("/")[0])>=2010 and int(x[1].split("/")[0])<=2020) \
    .map(lambda x: (x[0],1)).reduceByKey(lambda a,b: a+b)


cleanedUsersRDD=usersRDD.map(lambda x: (x.split(",")[0],(x.split(",")[3],x.split(",")[4]))) \


joinedRDD=cleanedPurchasesRDD.join(cleanedUsersRDD).filter(lambda x: x[1][1][0]=='IT') \
    .map(lambda x: (x[1][1][1],x[1][0])).reduceByKey(lambda a,b: a+b)

joinedRDD.saveAsTextFile(outputPath1)
```

#Task 2

```python
outputPath2="./output2"

def returnItems(row):
  key,year=row
  item23=0
  flag=0
  if year=='2023':
    item23+=1
  elif year=='2024':
    flag+=1

  return (key,(item23,flag))



purchases23RDD=purchasesRDD.map(lambda x: (x.split(",")[1],x.split(",")[3].split("/")[0])) \
    .map(returnItems).reduceByKey(lambda v1,v2: (v1[0]+v2[0],v1[1]+v2[1])) \
    .filter(lambda x: x[1][1]==0).map(lambda x: (x[0],x[1][0]))

purchases23RDD.saveAsTextFile(outputPath2)
```
