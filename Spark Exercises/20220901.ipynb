```python
from pyspark import SparkConf, SparkContext
conf = SparkConf().setAppName("Prova_esame")
sc = SparkContext(conf=conf)
```

#Task 1

```python
dataCentersRDD=sc.textFile("./data/DataCenters.txt")
companiesRDD=sc.textFile("./data/Companies.txt")

def occurrencies(row):
    codC, sizes = row
    sizes_list = list(sizes)
    numCenters = len(sizes_list)
    if numCenters >= 10:
        return (codC, (min(sizes_list), max(sizes_list)))



cleanedCentersRDD=dataCentersRDD.map(lambda x: (x.split(",")[1],(int(x.split(",")[2]),x.split(",")[3],x.split(",")[5]))) \
    .filter(lambda x: x[1][2]=='Europe').distinct().map(lambda x: (x[0],x[1][0])).groupByKey().map(occurrencies).filter(lambda x: x is not None)

cleanedCompaniesRDD=companiesRDD.map(lambda x: (x.split(",")[0],(x.split(",")[1],int(x.split(",")[2])))) \
    .filter(lambda x: x[1][1]>=200).map(lambda x: (x[0],x[1][0]))

joinedRDD=cleanedCompaniesRDD.join(cleanedCentersRDD).map(lambda x: (x[0], x[1][0], x[1][1][0], x[1][1][1]))
```
