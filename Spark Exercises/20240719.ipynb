#Task 1

```python
inputPath1="./data/Lectures.txt"
outputPath1="./output1/"
```

```python
lecturesRDD=sc.textFile(inputPath1)
cleanedLecturesRDD=lecturesRDD.map(lambda x: (x.split(",")[1],x.split(",")[5])).groupByKey()

def discard(lectures):
  target=120
  targetProp=0.7
  totLec=0
  longLec=0
  for lecture in lectures:
    if int(lecture)>=target:
      longLec+=1
      totLec+=1
    else:
      totLec+=1
  if longLec/totLec>=targetProp:
    return 1
  else:
    return 0

longLecturesRDD=cleanedLecturesRDD.mapValues(discard).filter(lambda x: x[1]==1).keys()
longLecturesRDD.saveAsTextFile(outputPath1)
```

#Task 2

```python
inputPath2="./data/UsersWatchedRecordedLectures.txt"
outputPath2="./output2/"
```

```python
UserRecordsRDD=sc.textFile(inputPath2)

def countMultiple(row):
  flag=0
  temp=[]
  for value in row[1]:
    if value in temp:
      flag +=1
    else:
      temp.append(value)


  if flag == 0:
    return (row[0][0],row[0][1])


cleanedRecordsRDD=UserRecordsRDD.map(lambda x: ((x.split(",")[0],x.split(",")[3]),x.split(",")[2])).groupByKey().map(countMultiple).filter(lambda x: x is not None)
cleanedRecordsRDD.saveAsTextFile(outputPath2)
```
